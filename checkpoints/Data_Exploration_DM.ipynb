## (C) Dataset Selection

### Selected Dataset: Airbnb United States Listings

**Reasons:**
- Directly supports frequent itemsets and association rules on `amenities` (course).
- Naturally supports clustering of listings by price, location, and capacity (course).
- Provides rich text (`description`, review `comments`) for text mining and embeddings (course).
- Enables beyond-course methods like XGBoost for price prediction (external).
- Enables topic modeling (LDA) and BERT-based sentiment on descriptions/reviews (external).
- Includes simple graph structure (host–listing) for basic graph mining and potential node embeddings (course/external).
- Multi-city structure allows hierarchical or mixed-effects modeling with city as a higher level (external).

**Trade-offs:**
- Graph structure is relatively simple compared to GitHub; less suitable for advanced GNN experiments.
- No natural long sequential behavior like playlists, so sequence models are less central than with Spotify MPD.
- City and market heterogeneity must be explicitly modeled (city effects), or results can be misleading.
- Housing and pricing are ethically sensitive topics (gentrification, affordability), so analysis must be aggregated and carefully framed.

## (D) Exploratory Data Analysis

````python
from bs4 import BeautifulSoup
from google.colab import files
````

````python
import requests # Import the requests library to fetch content from URLs

html_url = "https://insideairbnb.com/get-the-data/" # <<< REPLACE THIS WITH YOUR TARGET URL

extracted_urls_listings = []
extracted_urls_reviews = []

print(f"Processing URL: {html_url}")

try:
    # Fetch content from the URL
    response = requests.get(html_url)
    response.raise_for_status() # Raise an HTTPError for bad responses (4xx or 5xx)
    content = response.content

    # BeautifulSoup can handle bytes directly and detect encoding
    soup = BeautifulSoup(content, 'html.parser')

    for link in soup.find_all('a', href=True):
        href = link['href']
        # Filter for URLs containing '/united-states/' and ending with 'listings.csv.gz'
        if '/united-states/' in href and href.endswith('listings.csv.gz'):
            extracted_urls_listings.append(href)
        if '/united-states/' in href and href.endswith('reviews.csv.gz'):
            extracted_urls_reviews.append(href)

    print(f"\nExtracted {len(extracted_urls_listings)} listings URLs:")
    for url in extracted_urls_listings:
        print(url)
    print(f"\nExtracted {len(extracted_urls_reviews)} reviews URLs:")
    for url in extracted_urls_reviews:
        print(url)
except requests.exceptions.RequestException as e:
    print(f"Error fetching URL {html_url}: {e}")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
````

````python
import pandas as pd

# --- Process Listings ---
listings_dfs = []
print(f"Downloading and combining {len(extracted_urls_listings)} listing files...")

for url in extracted_urls_listings:
    print(f"Loading: {url}")
    try:
        # pandas automatically detects gzip compression from the file extension
        df = pd.read_csv(url)
        # Optionally add a column to track source if needed, e.g., city/url
        # df['source_url'] = url
        listings_dfs.append(df)
    except Exception as e:
        print(f"Error loading {url}: {e}")

if listings_dfs:
    listings_df = pd.concat(listings_dfs, ignore_index=True)
    print(f"\nSuccessfully combined listings. Shape: {listings_df.shape}")
    display(listings_df.head())
else:
    print("No listings data could be loaded.")

# --- Process Reviews ---
reviews_dfs = []
print(f"\nDownloading and combining {len(extracted_urls_reviews)} review files...")

for url in extracted_urls_reviews:
    print(f"Loading: {url}")
    try:
        df = pd.read_csv(url)
        reviews_dfs.append(df)
    except Exception as e:
        print(f"Error loading {url}: {e}")

if reviews_dfs:
    reviews_df = pd.concat(reviews_dfs, ignore_index=True)
    print(f"\nSuccessfully combined reviews. Shape: {reviews_df.shape}")
    display(reviews_df.head())
else:
    print("No reviews data could be loaded.")
````

````python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

sns.set(style="whitegrid")

df = listings_df.copy()

print("=== BASIC INFO ===")
print(df.shape)
print(df.dtypes)

print("\n=== MISSING VALUES (count and %) ===")
missing_counts = df.isna().sum()
missing_pct = (missing_counts / len(df) * 100).round(2)
missing_summary = pd.DataFrame({
    "missing_count": missing_counts,
    "missing_pct": missing_pct
}).sort_values("missing_count", ascending=False)
display(missing_summary)
````

````python
# ---------------------------------------------------------------------------
# 1. Clean PRICE and Summarize
# ---------------------------------------------------------------------------

print("\n=== PRICE SUMMARY ===")
# Handle string prices like "$150.00"
if df["price"].dtype == "object":
    df["price_clean"] = (
        df["price"]
        .astype(str)
        .str.replace("$", "", regex=False)
        .str.replace(",", "", regex=False)
    )
    df["price_clean"] = pd.to_numeric(df["price_clean"], errors="coerce")
else:
    df["price_clean"] = df["price"].astype(float)

# Histogram of raw prices
plt.figure(figsize=(8,4))
sns.histplot(df["price_clean"].dropna(), bins=100, kde=False)
plt.xlim(0, df["price_clean"].quantile(0.99))  # cap at 99th percentile for visibility
plt.title("Price Distribution (Capped at 99th Percentile)")
plt.xlabel("Price (USD)")
plt.ylabel("Count")
plt.show()

# Histogram of log-transformed prices
plt.figure(figsize=(8,4))
sns.histplot(np.log1p(df["price_clean"].dropna()), bins=60, kde=True)
plt.title("Log(1 + Price) Distribution")
plt.xlabel("log(1 + price)")
plt.ylabel("Count")
plt.show()
````

````python
# ---------------------------------------------------------------------------
# 2. Room Type, City, and Capacity
# ---------------------------------------------------------------------------

print("\n=== ROOM TYPE DISTRIBUTION ===")
display(df["room_type"].value_counts())
print("\nROOM TYPE (%)")
display((df["room_type"].value_counts(normalize=True) * 100).round(2))

print("\n=== CITY DISTRIBUTION (TOP 10) ===")
display(df["neighbourhood_cleansed"].value_counts().head(10))

print("\n=== ACCOMMODATES SUMMARY ===")
display(df["accommodates"].describe().to_frame("accommodates"))
````

````python
# ---------------------------------------------------------------------------
# 3. Ratings Summary
# ---------------------------------------------------------------------------

print("\n=== REVIEW SCORES RATING ===")
display(df["review_scores_rating"].describe().to_frame("review_scores_rating"))
miss = df["review_scores_rating"].isna().sum()
print(f"Missing ratings: {miss} ({miss / len(df) * 100:.2f}%)")
````

````python
# ---------------------------------------------------------------------------
# 4. Amenities: Basket Size and Top Amenities
# ---------------------------------------------------------------------------

print("\n=== AMENITIES EDA ===")

def parse_amenities(x):
    if pd.isna(x):
        return []
    s = str(x).strip()
    s = s.strip("{}[]")
    parts = [p.strip().strip('"').strip("'") for p in s.split(",")]
    return [p for p in parts if p]

df["amenities_list"] = df["amenities"].apply(parse_amenities)
df["amenity_count"] = df["amenities_list"].apply(len)

print("\nAmenity count (basket size) summary:")
display(df["amenity_count"].describe().to_frame("amenity_count"))

all_amenities = pd.Series([a for lst in df["amenities_list"] for a in lst])
top_amenities = all_amenities.value_counts().head(20)
print("\nTop 20 amenities:")
display(top_amenities.to_frame("count"))

plt.figure(figsize=(6,4))
sns.scatterplot(
    data=df.sample(min(5000, len(df)), random_state=42),  # sample for speed
    x="accommodates",
    y="price_clean",
    alpha=0.3
)
plt.ylim(0, df["price_clean"].quantile(0.99))
plt.title("Price vs. Accommodates (Sample, Capped at 99th Percentile)")
plt.xlabel("Accommodates")
plt.ylabel("Price (USD)")
plt.show()
````

````python
# ---------------------------------------------------------------------------
# 5. Spatial Summary by City and Neighbourhood
# ---------------------------------------------------------------------------

print("\n=== PRICE BY NEIGHBOURHOOD (TOP 10 BY COUNT) ===")
neigh_counts = df["neighbourhood_cleansed"].value_counts()
top_neigh = neigh_counts.head(10).index

price_by_neigh = (
    df[df["neighbourhood_cleansed"].isin(top_neigh)]
    .groupby("neighbourhood_cleansed")["price_clean"]
    .agg(["count", "mean", "median"])
    .sort_values("count", ascending=False)
)
display(price_by_neigh)
````

## (E) Initial Insights and Direction

### Key Observations

**Observation 1 – Heavy price skew and extreme variance**  
- Price is highly skewed with a very long right tail (99th percentile near 1,000+ USD, max >5,000+ in the plots).  
- Median prices in many neighbourhoods are modest (≈150–220 USD), but means are much higher (e.g., Primary Urban Center mean ≈1,758 vs. median 159).  

**Observation 2 – Dense but varied “basket sizes” (amenities)**  
- Listings have many amenities: median 39 amenities, with some up to 138.  
- A small set of “safety/comfort” amenities (Smoke alarm, Wifi, Kitchen, CO alarm, Hot water) appears in most listings, while luxury amenities are much rarer.  

**Observation 3 – High rating compression and missingness**  
- Ratings are very high and compressed (mean ≈4.78, 75% at 5.0), with ≈20.7% listings missing ratings.  
- This makes it hard to distinguish quality differences using ratings alone, and introduces non-random missingness (new/low-traffic listings).  

**Observation 4 – Strong geographic effects on price**  
- Neighbourhoods like “Primary Urban Center”, “Lahaina”, “North Kona” show extremely high mean prices but more moderate medians, indicating pockets of ultra-expensive listings.  
- Other neighbourhoods (e.g., “San Jose”, “Hennepin”) have much lower average prices.  

**Observation 5 – Strong host concentration (“professionalization”)**  
- Most hosts have 1 listing (median host_listing_count = 1), but some hosts manage thousands (max 4,383).  
- Over half of listings (≈55.2%) belong to “professional” hosts (3+ listings).  

**Observation 6 – Amenities and capacity structure**  
- Amenity count is positively related to accommodates (corr ≈ 0.34), and both relate weakly to price and ratings.  
- Description length is substantial (median ≈ 439 characters), suggesting rich text for topic/sentiment modeling.

### Hypotheses

1. **Non-linear and heterogeneous pricing**  
   - *Hypothesis H1:* Non-linear models (e.g., XGBoost) on log(price) will substantially outperform linear models due to strong skew and geography-driven heterogeneity.  
   - *Test:* Compare baseline linear regression vs. XGBoost on log(price) with cross-validation.

2. **Amenity patterns vs. price and professionalism**  
   - *Hypothesis H2:* Certain amenity bundles (e.g., safety + luxury amenities) are strongly associated with high prices and are more common among professional hosts.  
   - *Test:* Run frequent itemset/association rule mining on `amenities_list`, then correlate rules with price and host type.

3. **Ratings saturation vs. text sentiment**  
   - *Hypothesis H3:* Textual features (topics/sentiment in descriptions and reviews) explain more variation in perceived quality than numeric ratings alone, especially within the 4.7–5.0 band.  
   - *Test:* Build models predicting high/low quality using ratings vs. text-based features (TF‑IDF, sentiment, topics).

4. **Professional vs. casual host behavior**  
   - *Hypothesis H4:* Professional hosts set higher prices (controlling for neighbourhood and accommodates) and offer more amenities but may not always achieve higher ratings.  
   - *Test:* Compare distributions of price, amenity_count, and review_scores_rating between professional and casual hosts, controlling for neighbourhood.

5. **Location-driven segmentation**  
   - *Hypothesis H5:* Clustering on [log(price), accommodates, amenity_count, latitude, longitude] produces geographically coherent “market segments” (e.g., resort coastal, urban budget, suburban family).  
   - *Test:* Run K-Means, examine cluster summaries and neighbourhood composition.

### Potential Research Questions (RQs)

**RQ1 – Price modeling and non-linearity**  
- How much does XGBoost improve price prediction over linear regression when using log(price) and including geography and host features?

**RQ2 – Amenity bundles and value**  
- Which amenity combinations are most strongly associated with high prices or high ratings?  
- Do professional hosts systematically use different amenity bundles than casual hosts?

**RQ4 – Professionalization and fairness**  
- How concentrated is the market in terms of host_listing_count, and how does this relate to price, occupancy proxies (e.g., reviews_per_month, if added), and ratings?  
- Are there systematic differences in amenities and pricing strategies between professional and casual hosts that might affect guest and local community outcomes?

**RQ5 – Geographic segmentation**  
- What listing clusters emerge when considering both location and features (price, capacity, amenities), and how do these clusters differ by neighbourhood and price level?  
- Do “resort” clusters (e.g., Lahaina, North Kona) show different pricing dynamics than “urban” clusters (e.g., Hollywood, Primary Urban Center)?
